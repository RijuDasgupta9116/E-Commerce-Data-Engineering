```python
# ============================================================
# Cell 2: Imports
# ============================================================
import uuid
from datetime import datetime, timezone
from pyspark.sql.types import StructType, StructField, StringType, LongType, TimestampType, DoubleType
```

```python
# ============================================================
# Cell 3: Config
# ============================================================
table_name = "order_reviews"
source_path = "/Volumes/ecommerce_dev/raw_data/landing/olist_order_reviews.csv"
checkpoint_path = f"/Volumes/ecommerce_dev/raw_data/landing/_checkpoints/{table_name}/_checkpoint/"
target_table = f"ecommerce_dev.bronze.{table_name}"
```

```python
# ============================================================
# Cell 4: Init run tracking
# ============================================================
run_id = str(uuid.uuid4())
start_time = datetime.now(timezone.utc)
```

```python
# ============================================================
# Cell 5: Truncate existing corrupted Bronze table
# ============================================================
spark.sql(f"TRUNCATE TABLE {target_table}")
```

```python
# ============================================================
# Cell 6: Clear checkpoint so Autoloader reprocesses the file from scratch
# ============================================================
try:
    dbutils.fs.rm(checkpoint_path, recurse=True)
except Exception as e:
    if "does not exist" not in str(e):
        raise
```

```python
# ============================================================
# Cell 7: Re-ingest with corrected CSV parsing options
# ============================================================
try:
    # Explicit schema — all string, matching Bronze convention (casting happens at Silver)
    schema = StructType([
        StructField("review_id", StringType(), True),
        StructField("order_id", StringType(), True),
        StructField("review_score", StringType(), True),
        StructField("review_comment_title", StringType(), True),
        StructField("review_comment_message", StringType(), True),
        StructField("review_creation_date", StringType(), True),
        StructField("review_answer_timestamp", StringType(), True)
    ])

    df = (spark.readStream
          .format("cloudFiles")
          .option("cloudFiles.format", "csv")
          .option("cloudFiles.useNotifications", "false")
          .option("pathGlobFilter", "olist_order_reviews.csv")
          .option("header", "true")
          .option("multiLine", "true")
          .option("quote", '"')
          .option("escape", '"')
          .option("rescuedDataColumn", "_rescued_data")
          .schema(schema)
          .load("/Volumes/ecommerce_dev/raw_data/landing/")
          .withColumn("_source_file", col("_metadata.file_path"))
          .withColumn("_source_table", lit(table_name))
          .withColumn("_ingested_at", current_timestamp())
         )

    query = (df.writeStream
             .format("delta")
             .option("checkpointLocation", checkpoint_path)
             .trigger(availableNow=True)
             .toTable(target_table))
    query.awaitTermination()

    rows_ingested = spark.table(target_table).count()
    status = "SUCCESS"
    error_message = None

except Exception as e:
    rows_ingested = None
    status = "FAILED"
    error_message = str(e)
    raise

finally:
    end_time = datetime.now(timezone.utc)
    duration = (end_time - start_time).total_seconds()

    log_schema = StructType([
        StructField("run_id", StringType(), True),
        StructField("layer", StringType(), True),
        StructField("table_name", StringType(), True),
        StructField("source_path", StringType(), True),
        StructField("status", StringType(), True),
        StructField("rows_ingested", LongType(), True),
        StructField("start_time", TimestampType(), True),
        StructField("end_time", TimestampType(), True),
        StructField("duration_seconds", DoubleType(), True),
        StructField("error_message", StringType(), True),
        StructField("triggered_by", StringType(), True),
        StructField("metadata", StringType(), True)
    ])

    log_entry = spark.createDataFrame([(
        run_id,
        "bronze",
        table_name,
        source_path,
        status,
        rows_ingested,
        start_time,
        end_time,
        duration,
        error_message,
        "riju9826@gmail.com",
        '{"file_format": "csv", "schema_evolution_mode": "addNewColumns", "target_table": "' + target_table + '", "fix_applied": "multiLine_quote_escape_reingest"}'
    )], schema=log_schema)
    log_entry.write.mode("append").saveAsTable("ecommerce_dev.audit.ingestion_log")
```

```python
# ============================================================
# Cell 8: Verify the fix
# ============================================================
df_check = spark.table(target_table)
display(df_check.groupBy("review_score").count().orderBy("review_score"))
print(f"Total rows: {df_check.count()}")
print(f"Rows with _rescued_data populated: {df_check.filter(col('_rescued_data').isNotNull()).count()}")
```
